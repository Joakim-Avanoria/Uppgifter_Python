### Uppgift 17 – Budget-sammanställning (CSV→JSON)
Skapa en CSV `transactions.csv` med kolumner `date,category,amount` (positivt för intäkt, negativt för kostnad).
Skriv kod som:
- läser transaktionerna,
- summerar **belopp per kategori** i en dictionary,
- skriver ut en **lista** med `{ "category": "...", "total": ... }` sorterad fallande på `total`,
- sparar listan till `budget_summary.json`.

````python
import csv
import json

rows = [
    ["2025-10-01", "Lön", 30000],
    ["2025-10-02", "Mat", -450],
    ["2025-10-03", "Transport", -120],
    ["2025-10-04", "Mat", -300],
]
with open("transactions.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["date", "category", "amount"])
    writer.writerows(rows)

totals = {}
with open("transactions.csv", "r", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        category = row["category"]
        totals[category] = totals.get(category, 0) + float(row["amount"])

summary = [
    {"category": category, "total": total}
    for category, total in totals.items()
]
summary.sort(key=lambda item: item["total"], reverse=True)
print(summary)

with open("budget_summary.json", "w", encoding="utf-8") as file:
    json.dump(summary, file, ensure_ascii=False, indent=2)
````

### Uppgift 23 – Arv (Inheritance)
Skapa en basklass `Shape` och underklasser som t.ex: `Rectangle(w,h)`, `Circle(r)` och `Triangle(base, height)`.
- Skapa minst 3 underklasser.
- Alla underklasser ska ha en metod `area()` (för cirkel: `3.14159 * r * r`).
- Skapa minst fyra objekt sammanlagt, med en blandning av de olika underklasserna.
- Lägg objekten i en lista och skriv ut klassnamn och area för varje objekt.

````python
class Shape:
    def area(self):
        raise NotImplementedError

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius * self.radius

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return self.base * self.height / 2

shapes = [Rectangle(3, 4), Circle(2), Triangle(5, 3), Rectangle(2, 8)]
for shape in shapes:
    print(type(shape).__name__, round(shape.area(), 2))
````

### Uppgift 24 – Bibliotekssystem (OOP)

Bygg ett litet bibliotekssystem med följande klasser:

- `Book(title, author)` representerar en bok.
- `Member(name)` representerar en medlem.
- `Library` hanterar bibliotekets böcker, medlemmar och utlåningar.

`Library` ska innehålla:

- en lista med bibliotekets böcker,
- en lista med registrerade medlemmar,
- en dictionary `loans` som håller reda på vem som har lånat vilken bok.

Låt `loans` använda bokens titel som nyckel och ett `Member`-objekt som värde. Du kan förutsätta att varje boktitel är unik i det här systemet.

Klassen `Library` ska erbjuda metoderna:

- `add_book(book)` – lägger till en bok,
- `add_member(member)` – registrerar en medlem,
- `borrow_book(member, title)` – lånar ut en tillgänglig bok,
- `return_book(member, title)` – lämnar tillbaka en bok,
- `show_loans()` – visar vilka böcker som är utlånade och till vem.

Programmet ska förhindra att:

- en bok som inte finns lånas ut,
- en redan utlånad bok lånas ut igen,
- fel medlem lämnar tillbaka en bok.

Håll all information i minnet. Inga filer behöver användas i denna uppgift.

````python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

class Member:
    def __init__(self, name):
        self.name = name

class Library:
    def __init__(self):
        self.books = []
        self.members = []
        self.loans = {}  # titel -> Member-objekt

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def borrow_book(self, member, title):
        if member not in self.members:
            return False
        if not any(book.title == title for book in self.books):
            return False
        if title in self.loans:
            return False
        self.loans[title] = member
        return True

    def return_book(self, member, title):
        if self.loans.get(title) is not member:
            return False
        del self.loans[title]
        return True

    def show_loans(self):
        if not self.loans:
            print("Inga böcker är utlånade.")
        for title, member in self.loans.items():
            print(f"{title} är utlånad till {member.name}")

library = Library()
book = Book("Python 101", "A. Författare")
member = Member("Clara")
other_member = Member("David")
library.add_book(book)
library.add_member(member)
library.add_member(other_member)
print(library.borrow_book(member, "Python 101"))
print(library.borrow_book(other_member, "Python 101"))
library.show_loans()
print(library.return_book(other_member, "Python 101"))
print(library.return_book(member, "Python 101"))
````

### Projekt 1 – Kursregister (CSV + OOP)
- Skapa en CSV `enrollments.csv` med kolumner `student,course,points`.
- Skapa en klass `Student` med attributen `name` och `points`.
- Skapa en klass `Course` med `name` och en lista med studenter.
- Läs CSV-filen rad för rad. Skapa ett `Student`-objekt för varje rad och lägg objektet i rätt `Course`.
- Skriv ut antal studenter och genomsnittligt antal poäng för varje kurs.
- Spara en sammanfattning till `course_summary.json` (lista med `{course, n_students, avg_points}`).

````python
import csv
import json

with open("enrollments.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["student", "course", "points"])
    writer.writerows([
        ["Ada", "Python", 85],
        ["Bob", "Python", 72],
        ["Clara", "Statistik", 91],
        ["David", "Statistik", 78],
    ])

class Student:
    def __init__(self, name, points):
        self.name = name
        self.points = points

class Course:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

courses = {}
with open("enrollments.csv", "r", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        course = courses.setdefault(row["course"], Course(row["course"]))
        course.add_student(Student(row["student"], float(row["points"])))

summary = []
for course in courses.values():
    average = sum(student.points for student in course.students) / len(course.students)
    summary.append({
        "course": course.name,
        "n_students": len(course.students),
        "avg_points": round(average, 1),
    })
    print(course.name, len(course.students), round(average, 1))

with open("course_summary.json", "w", encoding="utf-8") as file:
    json.dump(summary, file, ensure_ascii=False, indent=2)
````

### Projekt 2 – Ordfrekvens-pipeline (fil → dict → JSON)
- Läs en textfil, normalisera (små bokstäver, ta bort `, . ! ?`), räkna ord i en dictionary.
- Skriv ut **topp 10** mest frekventa orden.
- Spara hela frekvens-dictionaryn till `word_freq.json`.
- (Valfritt) Läs tillbaka JSON och skriv ut ord som förekommer **exakt** `N` gånger (från input).

````python
import json
from pathlib import Path

Path("text.txt").write_text(
    "Python är roligt. Python används för data, statistik och mer data!",
    encoding="utf-8",
)
text = Path("text.txt").read_text(encoding="utf-8").lower()
for character in ",.!?":
    text = text.replace(character, "")

frequencies = {}
for word in text.split():
    frequencies[word] = frequencies.get(word, 0) + 1

top_ten = sorted(frequencies.items(), key=lambda item: item[1], reverse=True)[:10]
print(top_ten)

with open("word_freq.json", "w", encoding="utf-8") as file:
    json.dump(frequencies, file, ensure_ascii=False, indent=2)


target_count = 2
print([word for word, count in frequencies.items() if count == target_count])
````
