### Uppgift 2 – Strängmanipulation
Läs in en mening och skriv ut den på tre sätt:
- utan vokaler
- utan konsonanter
- med orden i omvänd ordning  

Mellanslag, siffror och skiljetecken behöver inte tas bort. När orden skrivs i omvänd ordning ska bokstäverna i varje ord behålla sin vanliga ordning.

```python
text = input("Skriv en mening: ")
vowels = "aouåeiyäöAOUÅEIYÄÖ"

without_vowels = ""
without_consonants = ""
for character in text:
    if character not in vowels:
        without_vowels += character
    if character in vowels or not character.isalpha():
        without_consonants += character

reversed_words = " ".join(text.split()[::-1])
print("Utan vokaler:", without_vowels)
print("Utan konsonanter:", without_consonants)
print("Omvänd ordning:", reversed_words)
```

### Uppgift 3 – Lista med tal
Låt användaren mata in en lista med tal (separerade med kommatecken, t.ex. `1,4,7,3`).
Programmet ska sedan:
- räkna ut summan
- hitta minsta och största talet
- räkna ut medelvärdet
(Använd loopar, inte `min`, `max`, `sum`).

```python
raw = input("Skriv tal separerade med kommatecken: ")
numbers = [float(part.strip()) for part in raw.split(",")]

total = 0
smallest = numbers[0]
largest = numbers[0]
for number in numbers:
    total += number
    if number < smallest:
        smallest = number
    if number > largest:
        largest = number

print("Summa:", total)
print("Minsta:", smallest)
print("Största:", largest)
print("Medelvärde:", total / len(numbers))
```

### Uppgift 5 – Primtal
Skriv en funktion `is_prime(n)` som returnerar `True` om `n` är ett primtal, annars `False`. Ett primtal är ett heltal större än 1 som endast är delbart med 1 och sig självt.

```python
def is_prime(n):
    if n < 2:
        return False
    for divisor in range(2, int(n ** 0.5) + 1):
        if n % divisor == 0:
            return False
    return True

print(is_prime(2))
print(is_prime(9))
print(is_prime(11))
```

### Uppgift 7 – FizzBuzz som funktion
Skriv om din gamla FizzBuzz-kod från vecka 1 till en funktion `fizzbuzz(n)` som returnerar en lista med resultaten.

```python
def fizzbuzz(n):
    results = []
    for number in range(1, n + 1):
        if number % 15 == 0:
            results.append("FizzBuzz")
        elif number % 3 == 0:
            results.append("Fizz")
        elif number % 5 == 0:
            results.append("Buzz")
        else:
            results.append(number)
    return results

print(fizzbuzz(16))
```

### Uppgift 8 – Miniräknare
Skriv ett program där varje räknesätt är en funktion: `add(a,b)`, `sub(a,b)`, `mul(a,b)`, `div(a,b)`.  
Programmet ska visa en meny där användaren väljer operation och matar in två tal. Programmet ska också hantera ett försök att dividera med noll utan att krascha.

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    if b == 0:
        return None
    return a / b

while True:
    choice = input("Välj +, -, *, / eller q: ")
    if choice == "q":
        break
    a = float(input("Första talet: "))
    b = float(input("Andra talet: "))
    operations = {"+": add, "-": sub, "*": mul, "/": div}
    if choice not in operations:
        print("Ogiltigt val")
        continue
    result = operations[choice](a, b)
    if result is None:
        print("Det går inte att dividera med noll.")
    else:
        print(result)
```

### Uppgift 12 – Telefonbok
Skriv ett menyprogram där användaren kan:
- lägga till en kontakt (namn + nummer)
- söka efter ett namn
- visa alla kontakter
- ta bort en kontakt
- avsluta programmet

Kontakterna sparas i en dictionary `{namn: nummer}`.

```python
phonebook = {}
while True:
    print("1 Lägg till | 2 Sök | 3 Visa alla | 4 Ta bort | 0 Avsluta")
    choice = input("Val: ")
    if choice == "1":
        name = input("Namn: ")
        phonebook[name] = input("Nummer: ")
    elif choice == "2":
        name = input("Namn: ")
        print(phonebook.get(name, "Ej hittad"))
    elif choice == "3":
        for name, number in phonebook.items():
            print(f"{name}: {number}")
    elif choice == "4":
        name = input("Namn: ")
        removed = phonebook.pop(name, None)
        print("Borttagen" if removed is not None else "Ej hittad")
    elif choice == "0":
        break
    else:
        print("Ogiltigt val")
```

### Projekt 2 – Sten-Sax-Påse: Turnering
Skapa ett program där man kan köra en turnering **Sten-Sax-Påse** mot datorn.

#### Krav:
* Representera valen som strängar: `"sten"`, `"sax"`, `"påse"`.
* Skriv en funktion `winner(p1, p2)` som returnerar `"p1"`, `"p2"` eller `"oavgjort"` enligt reglerna:
    * sten slår sax, sax slår påse, påse slår sten.
* Låt användaren spela flera rundor mot datorn tills man väljer att avsluta, eller gör färdiga "Bäst av 3/5/7"-lägen.
* Visa enkel poängställning med en dictionary, t.ex. {"spelare": 0, "dator": 0, "oavgjort": 0}.

#### Tips:
* Använd en dict för att definiera vilka drag som slår vilka, t.ex. `beats = {"sten": "sax", "sax": "påse", "påse": "sten"}`.
* Håll all input och utskrift i huvudflödet; låt funktionsdelarna vara “rena” och bara räkna ut svar.

```python
import random

beats = {"sten": "sax", "sax": "påse", "påse": "sten"}

def winner(p1, p2):
    if p1 == p2:
        return "oavgjort"
    if beats[p1] == p2:
        return "p1"
    return "p2"

score = {"spelare": 0, "dator": 0, "oavgjort": 0}
while True:
    player = input("sten, sax, påse eller q: ").lower()
    if player == "q":
        break
    if player not in beats:
        print("Ogiltigt val")
        continue
    computer = random.choice(list(beats))
    result = winner(player, computer)
    if result == "p1":
        score["spelare"] += 1
    elif result == "p2":
        score["dator"] += 1
    else:
        score["oavgjort"] += 1
    print("Datorn valde", computer)
    print(score)
```

### Projekt 3 – Hangman (ordgissning)
Programmet väljer slumpmässigt ett ord från en lista. Användaren gissar en bokstav i taget tills ordet är löst eller det maximala antalet felaktiga gissningar har uppnåtts.  

Krav:
- Visa vilka bokstäver som har hittats, exempelvis `py_h_n`.
- Spara redan gissade bokstäver.
- En bokstav som redan har gissats ska inte räknas som ett nytt försök.
- Användaren ska få göra högst sex felaktiga gissningar.
- Programmet ska tala om ifall användaren vann eller förlorade och visa det rätta ordet.

```python
import random
tries = 6

words = ["python", "omniway", "programmering"]
secret = random.choice(words)
guessed = set()

while tries > 0:
    status = "".join(char if char in guessed else "_" for char in secret)
    print(status)

    if status == secret:
        print("Grattis, du vann!")
        break

    guess = input("Gissa en bokstav: ").lower()
    
    if len(guess) != 1 or not guess.isalpha():
        print("Skriv exakt en bokstav.")
        continue

    if guess in guessed:
        print("Den bokstaven har du redan gissat.")
        continue


    guessed.add(guess)

    if guess not in secret:
        tries -= 1
        print(f"Fel. Du har {tries} försök kvar")
else:
    print("Du förlorade. Ordet var", secret)
```
